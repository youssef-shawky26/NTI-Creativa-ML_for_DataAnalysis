```python
import numpy as np
np.random.seed(42)
```

## Task 1 — Create the data (5 min)

Create a 5×6 array called `grades` of random integers between 40 and 100 (inclusive of 40,
exclusive of 101), representing grades for 5 students across 6 subjects.

Also create an array `students = np.array(['Layla', 'Omar', 'Sara', 'Kareem', 'Nour'])` and
`subjects = np.array(['Math', 'Science', 'English', 'History', 'Art', 'PE'])`.

```python
# Your code here
grades = np.random.randint(40,101,size=(5,6))
students = np.array(['Layla', 'Omar', 'Sara', 'Kareem', 'Nour'])
subjects = np.array(['Math', 'Science', 'English', 'History', 'Art', 'PE'])

grades
```

## Task 2 — Curve the grades (10 min)

Any grade **below 50** should get a **+5 curve**. Do this with boolean masking (no loops).
Print the updated `grades` array.

```python
# Your code here
grades[grades<50] +=5
grades
```

## Task 3 — Row and column statistics (10 min)

Using the correct `axis`, compute:
- `student_avg`: each student's average grade across all subjects
- `subject_avg`: each subject's average grade across all students

Print both, labeled with the corresponding names from `students` / `subjects`.

```python
# Your code here
student_avg = grades.mean(axis=1) 
subject_avg = grades.mean(axis=0)

print(student_avg)
print(subject_avg)
```

## Task 4 — Who's on top? (10 min)

Using `np.argmax` / `np.argmin` (not manual comparison):
- Find the name of the **top-performing student** (highest `student_avg`).
- Find the name of the **hardest subject** (lowest `subject_avg`).

```python
# Your code here
top= np.argmax(student_avg); students[top]
low= np.argmin(subject_avg); subjects[low]

print(top)
print(low)
```

## Task 5 — Pass/Fail matrix (10 min)

Using `np.where`, build a `status` array the same shape as `grades`, containing `'Pass'`
where the grade is ≥ 60 and `'Fail'` otherwise.

Then count how many total "Fail" entries exist in the whole matrix.

```python
# Your code here

status = np.where(grades >= 60, 'Pass', 'Fail')
fail_count = np.sum(status == 'Fail')

print("Status matrix:\n", status)
print("Total fails:", fail_count)
```

## Task 6 — A new subject arrives (10 min)

A 6th subject, Music, is added. Generate a random grade for each of the 5 students
(same range as before) and add it as a **new column** to `grades` (use `np.append`,
`np.hstack`, or `np.column_stack` — your choice). Don't forget to also extend `subjects`
with `'Music'`.

Recompute `student_avg` with the new data.

```python
# Your code here
g_mus = np.random.randint(40,101,size=(5,1))
grades = np.column_stack([grades,g_mus])

student_avg= grades.mean(axis=1)

student_avg
```

## Task 7 — Rank the students (10 min)

Use `np.argsort` on the updated `student_avg` to print students **from highest to lowest**
average grade (name + average, formatted nicely).

> Hint: `np.argsort` sorts ascending by default — you'll need to reverse it.

```python
# Your code here




p = np.argsort(student_avg)
print(p)
print(students[p],": ",student_avg[p])
```

## 🌟 Bonus — Normalize each student's grades (optional, ~10 min)

For each student (row), rescale their grades to a 0–1 range using min-max normalization:

`normalized = (grade - row_min) / (row_max - row_min)`

Do this **without a loop** — use `axis`-aware `.min(axis=...)` / `.max(axis=...)` together
with broadcasting (you'll need `keepdims=True` so the shapes line up for row-wise subtraction).

```python
# Your code here
row_min= grades.min(axis=1,keepdims=True)
row_max= grades.max(axis=1,keepdims=True)

normalized = (grades-row_min) / (row_max-row_min)
normalized
```
